# Lidando com Outliers

```python
%matplotlib inline

import numpy as np

# gero 10 mil rendas em volta de 27 mil
incomes = np.random.normal(27000, 15000, 10000)

# jogo um bilionario no meio pra baguncar
incomes = np.append(incomes, [1000000000])

import matplotlib.pyplot as plt

plt.hist(incomes, 50) # o grafico fica inutil por causa de um valor so

plt.show()
```

```python
print(incomes.mean()) # a media sobe muito por causa dele
```

```python
def reject_outliers(data):
    u = np.median(data) # uso a mediana porque ela nao se abala com o outlier
    s = np.std(data)    # desvio padrao

    # fico so com quem esta a menos de 2 desvios da mediana
    filtered = [e for e in data if (u - 2 * s < e < u + 2 * s)]

    return filtered

filtered = reject_outliers(incomes)

plt.hist(filtered, 50) # agora da pra enxergar a distribuicao de verdade
plt.show()
```

```python
print(np.mean(filtered)) # media limpa bem perto dos 27 mil que eu pedi
```
