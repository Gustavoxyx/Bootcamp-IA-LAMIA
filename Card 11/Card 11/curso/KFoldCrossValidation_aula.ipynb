# K-Fold Cross Validation

O curso importa de `sklearn.cross_validation` mas esse modulo nao existe mais e agora e `sklearn.model_selection`.

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score
from sklearn import datasets
from sklearn import svm

# uso a mesma base de flores do PCA
iris = datasets.load_iris()
```

```python
from sklearn.model_selection import train_test_split

# separo 40 por cento dos dados pro teste
X_train, X_test, y_train, y_test = train_test_split(
    iris.data,
    iris.target,
    test_size=0.4,
    random_state=0
)

# monto um SVC que separa as especies com uma reta
clf = svm.SVC(kernel='linear', C=1).fit(X_train, y_train)

clf.score(X_test, y_test)
```

```python
# agora em vez de uma divisao so eu testo 5 vezes trocando o pedaco de fora
scores = cross_val_score(
    clf,         # o modelo
    iris.data,   # os dados completos
    iris.target, # as respostas certas
    cv=5         # em quantos pedacos eu corto
)

print(scores)        # a nota de cada pedaco
print(scores.mean()) # a media das 5
```

```python
# troco a reta por uma curva pra ver se melhora
clf = svm.SVC(kernel='poly', C=1).fit(X_train, y_train)
scores = cross_val_score(
    clf,
    iris.data,
    iris.target,
    cv=5
)

print(scores)
print(scores.mean())
```

```python
clf = svm.SVC(kernel='poly', C=1).fit(X_train, y_train)

# volto pro teste simples so pra comparar com a media do K-Fold
clf.score(X_test, y_test)
```
