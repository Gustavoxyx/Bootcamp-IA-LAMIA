```python
np.arange(1, 10, 2)   # start entra, stop não entra, step é o tamanho do passo
                      # pra ver a doc da função uso shift + tab
```

```python
arr = np.zeros((5 ,5))   # crio uma matriz de zeros e passo o shape na tupla
arr
```

```python
np.ones((3, 3))   # o primeiro parâmetro é o shape
```

```python
np.full((3, 3), 20)   # o numpy só tem função pronta pro 0 e pro 1,
                      # pra qualquer outro valor eu uso o full
```

```python
np.eye(4)   # crio a identidade os 1 ficam na diagonal e o resto é 0
            # uso muito em inversão e equação matricial
```

```python
np.linspace(0, 10, 2)   # o terceiro número é quantos valores quero igualmente espaçados
```

```python
np.linspace(0, 10, 3)
```

```python
np.random.rand(5)   # o random é sub-biblioteca do np e serve pra sortear
                    # o rand devolve entre 0 e 1 com a mesma chance tendo uma distribuição uniforme
```

```python
np.random.rand(5) * 100   # multiplico pra esticar a faixa de 0 até 100
```

```python
np.random.randn(4)   # sorteio na distribuição gaussiana gira em torno do 0
```

```python
np.random.randint(0, 100, 10)   # de, até e quantidade. devolve inteiro
```

```python
np.round(np.random.rand(5)*100, 0)   # o 0 é o número de casas decimais,
                                     # o round arredonda pro mais próximo
```

```python
arr = np.random.rand(25)   # criei um vetor de 25 elementos
arr
```

```python
arr = arr.reshape((5,5))   # transformei o vetor de 25 numa matriz 5x5
arr
```

```python
arr.shape   # é característica e não função uso sem parêntese
```

```python
arr.max()   # encontro o maior valor do array
```

```python
arr.min()   # encontro o menor
```

```python
arr.argmax()   # aqui vem o índice do maior, não o valor
```

```python
arr.argmin()   # índice do menor
```
