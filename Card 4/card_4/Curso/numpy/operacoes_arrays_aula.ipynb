```python
import numpy as np
```

```python
arr = np.arange(0, 16)
arr
```

```python
arr + arr   # a soma acontece elemento por elemento
```

```python
arr - arr
```

```python
arr = arr   # essa linha não muda nada.
```

```python
arr / arr   # no numpy a divisão por 0 não dá erro ele avisa e troca o valor por nan
```

```python
1/0   # já no python puro a divisão por 0 quebra o programa
```

```python
1/arr   # dividir por zero aqui devolve infinito (inf)
```

```python
arr ** 2   # a potenciação ocorre em todos os elementos
```

```python
arr * 100   # a multiplicação ocorre em todos os elementos
```

```python
arr + 100   # a soma ocorre em todos os elementos
```

```python
arr - 100   # a subtração ocorre em todos os elementos
```

```python
np.sqrt(arr)   # raiz de cada elemento
```

```python
np.exp(arr)   # exponencial de cada elemento
```

```python
np.mean(arr)   # média de todos
```

```python
np.std(arr)   # desvio padrão
```

```python
np.sin(arr)   # seno de cada elemento
```

```python
np.max(arr)
```

```python
np.min(arr)
```
