```python
import numpy as np
```

```python
arr = np.arange(0, 30, 3)
arr
```

```python
arr[3]   # indexação é a forma mais rápida de pegar um elemento
```

```python
arr[:5]   # sem nada antes dos dois pontos ele pega todos até ali
```

```python
arr[2:]   # lógica contrária todos depois do n
```

```python
arr[2:] = 100   # atribuo o mesmo valor pra fatia inteira
```

```python
arr
```

```python
arr = np.arange(50).reshape((5,10))
arr
```

```python
arr.shape
```

```python
# pra pegar linha e coluna o certo é arr[:3, :] com vírgula no mesmo colchete
    # esse arr[:3][:] são dois cortes de LINHA EM SEQUENCIA sai igual só por coincidência
arr[:3][:]
```

```python
arr2 = arr[:3].copy()   # sem o copy ele só aponta pro arr[:3] em vez de copiar a cecula
```

```python
arr2[:] = 100
```

```python
arr   # não alterou porque o arr2 era uma cópia de verdade
```

```python
arr2
```

```python
arr[:4,5:]   # agora sim 4 primeiras linhas e as colunas da 5 pra frente
```

```python
bol = arr > 50   # comparo e recebo um array de True e False
```

```python
arr[bol]   # vem vazio porque o maior valor aqui é 49 nada passa de 50
```
