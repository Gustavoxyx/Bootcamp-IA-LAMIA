```python
import pandas as pd
```

```python
data = {'Empresa':['GOOG','GOOG','MSFT','MSFT','FB','FB'],
       'Nome':['Sam','Charlie','Amy','Vanessa','Carl','Sarah'],
       'Venda':[200,120,340,124,243,350]}
```

```python
df = pd.DataFrame(data)
df
```

```python
df.groupby('Empresa')   # ele faz o agrupamento que é basicamente falar por quem eu quero agrupar
                        # sozinho ele não calcula nada só me devolve o objeto
```

```python
group = df.groupby("Empresa")
```

```python
group.sum(numeric_only=True)   #essa versão disponibilizada no nosso curso hoje em dia não funciona
# nas versões mais novas dá erro se tiver coluna de texto junto então não dá pra somar nome
```

```python
group.describe()   # peço várias estatísticas de uma vez
```
