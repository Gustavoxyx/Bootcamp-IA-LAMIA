```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({'col1':[1,2,3,4],'col2':[444,555,666,444],'col3':['abc','def','ghi','xyz']})
df.head()
```

```python
df['col2'].unique()   # devolve apenas os valores únicos
```

```python
np.unique(df['col2'])   # o numpy faz o mesmo só que já devolve ordenado
```

```python
df['col2'].nunique()   # retorna o tamanho do vetor de resposta ou seja quantos únicos tem
```

```python
df['col2'].value_counts()   # conto quantas vezes cada valor aparece
```

```python
newdf = df[(df['col1']>2) & (df['col2']==444)]   # permite seleções com os colchetes
newdf
```

```python
df = pd.DataFrame({'col1':[1,2,3,np.nan],
                   'col2':[np.nan,555,666,444],
                   'col3':['abc','def','ghi','xyz']})
df.head()
```

```python
data = {'A':['foo','foo','foo','bar','bar','bar'],
     'B':['one','one','two','two','one','one'],
       'C':['x','y','x','y','x','y'],
       'D':[1,3,2,5,4,1]}

df = pd.DataFrame(data)
df
```

```python
df.pivot_table(values='D',index=['A', 'B'],columns=['C'])   # cruzo duas coisas: uma vira linha e a outra coluna
                                                            # em index passo lista daí saio com índice multinivel
                                                            # combinação que não existe nos dados vira nan
```
