```python
import numpy as np
import pandas as pd
from numpy.random import randn
```

```python
np.random.seed(101)
```

```python
df = pd.DataFrame(randn(5, 4), index='A B C D E'.split(), columns='W X Y Z'.split())
df
```

```python
bol = df>0   # condição na tabela toda
```

```python
df[bol]   # mantém o tamanho e troca por nan onde reprovou não some com linha
```

```python
df[df['W']>0]['Y']   # condição numa coluna só aí sim some com as linhas e pego a coluna Y
```

```python
bol = df['W']>0
df2 = df[bol]
df2['Y']   # mesma coisa da célula de cima só que em dois passos
```

```python
df[(df['W']>0) | (df['Y']>1)]   # uso | pra ou e & pra e, o parentese é obrigatório
```

```python
df.reset_index(inplace=True)   # redefine o idx para o padrão 0 1 2...
df
```
