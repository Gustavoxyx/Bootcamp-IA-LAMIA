```python
import pandas as pd
import numpy as np
```

```python
outside = ['G1', 'G1', 'G1', 'G2', 'G2', 'G2']   # nível de fora
inside = [1, 2, 3, 1, 2, 3]                       # nível de dentro
```

```python
hier_index = list(zip(outside,inside))   # monto os pares (G1,1), (G1,2) etc...
hier_index
```

```python
hier_index = pd.MultiIndex.from_tuples(hier_index)   # converto num índice de dois níveis
hier_index
```

```python
df  = pd.DataFrame(np.random.randn(6, 2), index=hier_index, columns=['A', 'B'])
df
```

```python
df.index.names = ['Grupo', 'Numero']   # dou nome pros níveis e fica bem mais legível
df
```

```python
df.xs(1, level='Numero')   # com o xs corto um nível direto sem encadear loc
```
