```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({'A':[1, 2, np.nan], 'B': [5, np.nan, np.nan], 'C':[1,2,3]})
df
```

```python
df.dropna()   # o padrão é axis=0 então se tem nan na linha ela é excluída
```

```python
df.dropna(axis=1)   # com axis=1 a coluna que tiver nan é excluída
```

```python
df.dropna(thresh=2)   # o thresh é o mínimo de valores bons pra linha ficar
                      # não é a quantidade de nan que faz ela sair
```

```python
df['A'].fillna(value=df['A'].mean())   # preencho com a média da própria coluna
```
