```python
registros = pd.DataFrame({
    'vendedor': ['Ana', 'Bruno', 'Ana', 'Carla', 'Bruno', 'Carla', 'Ana'],
    'produto':  ['tenis', 'caneta', 'mochila', 'tenis', 'mochila', 'caderno', 'caneta'],
    'qtd':      [12, 8, np.nan, 5, 13, 2, 20],
    'desconto': [0.1, np.nan, np.nan, 0.15, 0.05, np.nan, 0.2]
})
print(registros)
```

```python
print(registros.isnull().sum())
registros.info()
```

```python
print(registros.dropna())
print(registros.dropna(axis=1))
print(registros.dropna(thresh=4))   # mínimo de valores válidos para manter a linha
```

```python
registros['qtd'] = registros['qtd'].fillna(registros['qtd'].mean())
registros['desconto'] = registros['desconto'].fillna(0)   # ausência aqui significa sem desconto
print(registros.isnull().sum())
```

```python
serie = pd.Series([10.0, np.nan, np.nan, 40.0])
print(serie.ffill())
```

```python
print(registros['vendedor'].unique())
print(np.unique(registros['produto']))   # o numpy já devolve ordenado
print(registros['vendedor'].nunique())
print(registros['vendedor'].value_counts())
```

```python
registros['preco'] = registros['produto'].apply(lambda p: produtos[p])
registros['receita'] = registros['qtd'] * registros['preco'] * (1 - registros['desconto'])
print(registros.round(2))
```

```python
def faixa(valor):
    return 'alta' if valor > 200 else 'baixa'

registros['faixa'] = registros['receita'].apply(faixa)   # passo a função sem parêntese
print(registros[['vendedor', 'receita', 'faixa']].round(2))
```

```python
print(registros.sort_values(by='receita'))
print(registros.sort_values(by='receita', ascending=False))
print(registros['faixa'].value_counts())
```
