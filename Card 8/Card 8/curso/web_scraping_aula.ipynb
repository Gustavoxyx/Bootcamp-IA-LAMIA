```python
# ---- bibliotecas ----
import requests
from bs4 import BeautifulSoup as bs
```

```python

r = requests.get("https://keithgalli.github.io/web-scraping/example.html")   # pego o html da pagina

soup = bs(r.content)   # transformo em objeto soup pra dar pra navegar

print(soup.prettify())   # imprimo formatado com identacao
```

```python
# ---- find e find_all ----
first_header = soup.find("h2")   # pego so o primeiro h2 que aparece

headers = soup.find_all("h2")   # pego todos os h2 da pagina numa lista
print(headers)
```

```python
# passo uma lista de tags pra procurar mais de um tipo de uma vez
first_header = soup.find(["h1", "h2"])

headers = soup.find_all(["h1", "h2"])
headers
```

```python
# filtro por atributo tipo o id
paragraph = soup.find_all("p", attrs={"id": "paragraph-id"})
paragraph
```

```python
# busco em cascata vou isolando ate chegar na tag que eu quero
body = soup.find("body")     # pego o body inteiro
div = body.find('div')       # dentro do body pego a primeira div
header = div.find('h1')      # dentro da div pego o h1
header
```

```python
# busco so um pedaco do texto usando regex sem precisar saber o texto exato
import re

paragraphs = soup.find_all("p", string=re.compile("Some"))   # qualquer p que tenha "Some" no texto
paragraphs

headers = soup.find_all("h2", string=re.compile("(H|h)eader"))   # pego com H ou h minusculo
headers
```

```python
# ---- select (seletor css) ----
print(soup.body.prettify())
```

```python
content = soup.select("div p")   # pego os p que estao dentro de uma div
content
```

```python
paragraphs = soup.select("h2 ~ p")   # pego os p que vem logo depois de um h2 no mesmo nivel
paragraphs
```

```python
bold_text = soup.select("p#paragraph-id b")   # pego o negrito que ta dentro do paragrafo com esse id
bold_text
```

```python
# aqui uso o > pra pegar só o filho direto do body e não qualquer descendente
paragraphs = soup.select("body > p")
print(paragraphs)

# pra cada paragrafo que achei busco dentro dele as tags em italico
for paragraphs in paragraphs:   # reaproveitei o nome da variavel e tomar cuidado que isso confunde depois
    print(paragraphs.select("i"))
```

```python

from bs4 import BeautifulSoup as bs
r = requests.get("https://keithgalli.github.io/web-scraping/webpage.html")
webpage = bs(r.content)
print(webpage.prettify())
```

```python

# jeito 1: direto com select
links = webpage.select("ul.socials a")
actual_links = [link['href'] for link in links]
links
```

```python
# jeito 2: acho a lista primeiro com find, depois os links dentro dela
links = webpage.find("ul", attrs={"class": "socials"})
actual_links = [link['href'] for link in links.find_all("a")]
actual_links
```

```python
# jeito 3: seleciono direto o li que tem a classe social
links = webpage.select("li.social a")
actual_links = [link['href'] for link in links]
actual_links
```

```python

import pandas as pd

table = webpage.select("table.hockey-stats")[0]   # acho a tabela pela classe
columns = table.find("thead").find_all("th")       # pego os nomes das colunas
columns_names = [c.string for c in columns]

table_rows = table.find("tbody").find_all("tr")
l = []
for tr in table_rows:
    td = tr.find_all('td')
    row = [str(tr.get_text()).strip() for tr in td]   # tiro os espacos em branco
    l.append(row)

df = pd.DataFrame(l, columns=columns_names)
df.loc[df['Team'] != 'Did not play']   # filtro as linhas que nao interessam
```

```python
import re

# aqui eu digitei is em vez de li no seletor e o certo era "ul.fun-facts li"
# como não existe tag <is>, o select devolve lista vazia e o resto do codigo não acha nada
facts = webpage.select("ul.fun-facts is")
facts_with_is = [facts.find_all(string=re.compile("is")) for fact in facts]
facts_with_is = [fact.find_parent().get_text() for fact in facts_with_is if fact]
print(facts_with_is)
```

```python

import requests
from bs4 import BeautifulSoup as bs
url = "https://keithgalli.github.io//web-scraping/"
r = requests.get(url+"webpage.html")
webpage = bs(r.content)

images = webpage.select("div.row div.column img")   # acho a imagem
image_url = images[0]['src']                         # pego o caminho dela
full_url = url + image_url                            # monto o endereco completo

img_data = requests.get(full_url).content
with open('image_name.jpg', 'wb') as handler:
    handler.write(img_data)   # gravo a imagem na pasta local
```

```python

# pego os links de dentro do bloco de arquivos misteriosos
files = webpage.select("div.block a")
relative_files = [f['href'] for f in files]

url = "https://keithgalli.github.io//web-scraping/"
for f in relative_files:
    full_url = url + f
    page = requests.get(full_url)
    bs_page = bs(page.content)
    secret_word_element = bs_page.find("p", attrs={"id": "secret-word"})   # acho o id exato
    secret_word = secret_word_element.string
    print(secret_word)   # cada link solta uma palavra e no final eu junto e formo a frase
```
