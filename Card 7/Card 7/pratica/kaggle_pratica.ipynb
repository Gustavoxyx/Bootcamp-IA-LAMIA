Puxei a base Iris pelo botão Add Input do painel lateral.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/kaggle/input/datasets/uciml/iris/Iris.csv")
print(df.shape)
print(df.head())
```

```python
fig, ax = plt.subplots(figsize=(7, 4.2))

for especie, grupo in df.groupby("Species"):   # cada especie vira uma cor
    ax.scatter(grupo["SepalLengthCm"], grupo["PetalLengthCm"], label=especie, s=28)

ax.set_xlabel("comprimento da sépala (cm)")
ax.set_ylabel("comprimento da pétala (cm)")
ax.set_title("Iris: dispersão sépala x pétala")
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```
