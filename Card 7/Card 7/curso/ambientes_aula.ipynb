```python
import numpy as np #uso pra fazer os calculos
import matplotlib.pylab as plt #uso pra desenhar o grafico
import matplotlib #importa a lib toda meio que nem precisa mas deixei que nem no video
```

```python
t = np.arange(0.0, 2.0, 0.01) #crio um intervalo de tempo de 0 a 2 segundos pulando de 0.01 em 0.01
s = 1 + np.sin(2* np.pi * t) #calculo a tensao pra cada ponto

#monto o grafico
fig, ax = plt.subplots()
ax.plot(t,s)

ax.set(xlabel='time(s)', ylabel = 'voltagem (mV)',
        title = 'About as simple as it gets, folks')
#configuro os nomes dos eixos e o titulo
ax.grid()
#coloco uma grade no fundo que ajuda a ler os valores
```
