```python
import numpy as np
import matplotlib.pyplot as plt
# aqui nao precisei instalar nada, o colab ja vem com as duas
```

```python
x = np.linspace(0, 4 * np.pi, 400)   # gero 400 pontos de 0 ate duas voltas
y = np.sin(x)

fig, ax = plt.subplots(figsize=(7, 3.5))
ax.plot(x, y, color="#2e7d4f", linewidth=2)
ax.set_title("Gráfico de seno")
ax.set_xlabel("x")
ax.set_ylabel("sen(x)")
ax.grid(True, alpha=0.3)
plt.show()
```
