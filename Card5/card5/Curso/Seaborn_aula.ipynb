```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("http://media.sundog-soft.com/SelfDriving/FuelEfficiency.csv")   # baixo a base direto da web precisa de internet
gear_counts = df['# Gears'].value_counts()   # conto quantos carros tem cada numero de marchas

gear_counts.plot(kind='bar')   # esse grafico sai no estilo padrao do matplotlib
plt.show()
```

```python
import seaborn as sns
sns.set()   # so de chamar isso o seaborn ja muda o visual de tudo do grafico de matplotlib
```

```python
gear_counts.plot(kind='bar')   # mesmo grafico da celula anterior agora com o estilo novo
```

```python
df.head()   # olho as colunas antes de escolher o que cruzar
```

```python
sns.displot(df['CombMPG'])   # distribuicao de uma coluna so com histograma e curva
```

```python
df2 = df[['Cylinders', 'CityMPG', 'HwyMPG', 'CombMPG']]   # separo so as colunas numericas que interessam
df2.head()
```

```python
sns.pairplot(df2, hue = 'Cylinders', height=2.5);
# cruza todas as colunas duas a duas de uma vez
# uso isso pra achar onde existe relacao antes de calcular qualquer correlacao
# o hue pinta os pontos pelo numero de cilindros
```

```python
sns.scatterplot(x="Eng Displ", y = "CombMPG", data=df)   # motor maior consome mais da pra ver a tendencia
```

```python
sns.jointplot(x="Eng Displ", y="CombMPG", data=df)
# alem dos pontos ele desenha a distribuicao de cada eixo nas bordas
```

```python
sns.lmplot(x="Eng Displ", y = "CombMPG", data=df)   # o mesmo grafico com a reta de tendencia calculada
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(rc={'figure.figsize': (15, 5)})   # alargo a figura senao os nomes de fabricante se amontoam
ax = sns.boxplot(x='Mfr Name', y='CombMPG', data=df)   # uma caixa por fabricante
plt.xticks(rotation=45)   # giro os nomes pra caber
```

```python
ax = sns.swarmplot(x='Mfr Name', y='CombMPG', data=df)   # aqui vejo cada carro como um ponto
ax.set_xticklabels(ax.get_xticklabels(), rotation=45)   # mostra quantos modelos tem em cada faixa
```

```python
ax = sns.countplot(x='Mfr Name', data=df)   # conto quantos carros cada fabricante tem na base
ax.set_xticklabels(ax.get_xticklabels(), rotation=45)
```

```python
df2 = df.pivot_table(index='Cylinders', columns='Eng Displ', values='CombMPG', aggfunc='mean')
sns.heatmap(df2)   # a cor mostra o consumo medio no cruzamento de cilindros com tamanho do motor
# quadrado em branco e combinacao que nao existe na base
```
