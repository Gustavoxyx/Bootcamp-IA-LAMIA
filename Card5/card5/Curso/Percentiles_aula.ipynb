```python
import numpy as np
import matplotlib.pyplot as plt

vals = np.random.normal(0, 0.5, 10000)

plt.hist(vals, 50)   # a curva e simetrica entao ja espero o percentil 50 no zero
plt.show()
```

```python
np.percentile(vals, 50)   # percentil 50 e a propria mediana
```

```python
np.percentile(vals, 90)   # 90 por cento dos valores estao abaixo desse ponto
```

```python
np.percentile(vals, 20)   # comparo com o de cima pra enxergar a dispersao sem usar a media
```
