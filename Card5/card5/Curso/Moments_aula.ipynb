```python
import numpy as np
import matplotlib.pyplot as plt

vals = np.random.normal(0, 0.5, 10000)   # normal centrada no zero

plt.hist(vals, 50)   # olho o formato antes de medir os momentos
plt.show()
```

```python
np.mean(vals)   # primeiro momento, onde a curva esta centrada
```

```python
np.var(vals)   # segundo momento, o quanto ela se espalha
```

```python
import scipy.stats as sp
sp.skew(vals)   # terceiro momento pra que lado a cauda se estende
# deu perto de zero porque a curva e simetrica
```

```python
sp.kurtosis(vals)   # quarto momento o peso da cauda
# perto de zero quer dizer que nao tem excesso de valor extremo
```
