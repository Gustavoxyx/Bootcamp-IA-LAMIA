```python
from random import randint

import numpy as np   # importo o numpy pra gerar e medir os dados
```

```python
incomes = np.random.normal(27000, 15000, 10000)   # crio 10 mil rendas com media 27 mil e desvio 15 mil
```

```python
np.mean(incomes)   # tiro a media deve ficar perto dos 27 mil que pedi
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
plt.hist(incomes, 50)   # divido as rendas em 50 barras pra ver o formato
plt.show()   # o desenho fica no formato de sino
```

```python
np.median(incomes)   # o valor do meio bem perto da media porque a curva e simetrica
```

```python
incomes = np.append(incomes, [1000000000])   # jogo um bilionario no meio das 10 mil rendas
```

```python
np.mean(incomes)   # a media disparou por causa de um unico valor
```

```python
np.median(incomes)   # a mediana quase nao mexeu por isso confio mais nela com valor extremo
```
