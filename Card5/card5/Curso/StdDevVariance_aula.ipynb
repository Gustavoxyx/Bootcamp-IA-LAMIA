```python
import numpy as np
import matplotlib.pyplot as plt

incomes = np.random.normal(100.0, 20.0, 10000)   # pedi media 100 e desvio 20

plt.hist(incomes, 50)   # confiro no grafico se a dispersao bate com o que pedi
plt.show()
```

```python
incomes.std()   # o desvio sai perto de 20 na mesma unidade dos dados
```

```python
incomes.var()   # a variancia e o desvio ao quadrado entao cai perto de 400
# por isso prefiro ler o desvio que da pra comparar com os valores originais
```
