```python
from numpy import random
random.seed(0)   # fixo a semente pra dar sempre o mesmo resultado

totals = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0}   # quantas pessoas caem em cada faixa
purchases = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0}   # quantas dessas compraram
totalPurchases = 0

for _ in range(100000):
    ageDecade = random.choice([20, 30, 40, 50, 60, 70])   # sorteio a faixa etaria
    purchaseProbability = float(ageDecade) / 100.0   # a chance de compra cresce com a idade
    totals[ageDecade] += 1   # quem tem 20 compra 20 por cento das vezes
    if (random.random() < purchaseProbability):   # quem tem 70 compra 70 por cento
        totalPurchases += 1
        purchases[ageDecade] += 1
```

```python
totals   # as faixas ficaram parecidas porque o sorteio e uniforme
```

```python
purchases   # aqui sim cresce da faixa 20 pra 70 que e o efeito da idade
```

```python
totalPurchases   # o total de compras somando todas as faixas
```

```python
PEF = float(purchases[30]/float(totals[30]))
print('P(purschases | 30s): ' + str(PEF))   # divido as compras da faixa 30 pelo total dela
```

```python
PF = float(totals[30]/100000.0)
print("P(30's): " + str(PF))   # quanto a faixa 30 representa do total, perto de um sexto
```

```python
PE = float(totalPurchases)/100000.0
print("P(30's): " + str(PE))   # chance de comprar sem olhar a idade, bem maior que a da faixa 30
```

```python
print("P(30's, Purchase)" + str(float(purchases[30])/100000.0))   # ter 30 anos e comprar ao mesmo tempo
```

```python
print("P(30's)P(Purchase)" + str(PE*PF))
# multipliquei as duas probabilidades separadas
# se fossem independentes esse numero daria igual ao da celula anterior
# deu diferente entao idade e compra estao ligadas
```

```python
print((purchases[30] / 100000.0) / PF)   # volto pra P(compra dado 30) pelo caminho inverso
# bate com o PEF confirmando a conta
```

```python
from numpy import random
random.seed(0)

totals = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0}
purchases = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0}
totalPurchases = 0

for _ in range(100000):
    ageDecade = random.choice([20, 30, 40, 50, 60, 70])
    purchaseProbability = 0.4   # agora a chance e a mesma pra qualquer idade
    totals[ageDecade] += 1   # refiz o experimento so mudando essa linha
    if (random.random() < purchaseProbability):
        totalPurchases += 1
        purchases[ageDecade] += 1
```

```python
PEF = float(purchases[30]/float(totals[30]))
print('P(purschases | 30s): ' + str(PEF))   # chance de comprar entre os de 30
```

```python
PE = float(totalPurchases)/100000.0
print("P(30's): " + str(PE))
# os dois valores ficaram praticamente iguais
# saber a idade nao mudou a chance entao agora as variaveis sao independentes
```
