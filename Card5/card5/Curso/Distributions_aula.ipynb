```python
import numpy as np
import matplotlib.pyplot as plt

values = np.random.uniform(-10.0, 10.0, 100000)   # na uniforme toda faixa tem a mesma chance

plt.hist(values, 50)   # por isso o grafico sai reto e nao em sino
plt.show()
```

```python
from scipy.stats import norm
import matplotlib.pyplot as plt

x = np.arange(-3, 3, 0.001)   # crio o eixo com passo bem pequeno pra curva sair lisa
plt.plot(x, norm.pdf(x))   # uso pdf porque a variavel e continua
```

```python
import numpy as np
import matplotlib.pyplot as plt

mu = 5.0   # mu e onde a curva fica centrada
sigma = 2.0   # sigma e o quanto ela se espalha
values = np.random.normal(mu, sigma, 10000)

plt.hist(values, 50)   # o pico cai no 5 como pedi no mu
plt.show()
```

```python
from scipy.stats import expon
import matplotlib.pyplot as plt

x = np.arange(0, 10 , 0.001)   # comeco no zero porque a exponencial nao tem valor negativo
plt.plot(x, expon.pdf(x))   # ela despenca rapido serve pra tempo ate o proximo evento
```

```python
from scipy.stats import binom
import matplotlib.pyplot as plt

n, p = 10, 0.5   # 10 tentativas com 50 por cento de chance cada
x = np.arange(0, 10, 0.001)
plt.plot(x, binom.pmf(x, n, p))   # aqui e pmf e nao pdf porque contagem e valor discreto
```

```python
from scipy.stats import poisson
import matplotlib.pyplot as plt

mu = 500   # media de eventos esperados no intervalo
x = np.arange(400, 600, 0.5)
plt.plot(x, poisson.pmf(x, mu))   # tambem e pmf a poisson conta evento dentro de um intervalo
```
